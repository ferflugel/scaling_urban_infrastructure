```python
import seaborn as sns
import matplotlib.pyplot as plt

def zipf_distribution(N_cities, s):
    population, rank = [], []
    for i in range(N_cities):
        population.append(int(1000000 / (i + 1) ** s))
        rank.append(i + 1)
    return population, rank

plot_list = [[(0.60, 2500, 1.2), (0.75, 2500, 1.2), (0.90, 2500, 1.2)],
             [(0.60, 2500, 1.0), (0.75, 2500, 1.0), (0.90, 2500, 1.0)],
             [(0.60, 2500, 0.8), (0.75, 2500, 0.8), (0.90, 2500, 0.8)]]
```

```python
sns.set_style('white')
sns.set_palette('mako_r', 5)

for index, experiment in enumerate(plot_list[::-1]):
    p, r = zipf_distribution(experiment[0][1], experiment[0][2])
    sns.scatterplot(x=r, y=p, edgecolor=None, label=experiment[0][2])
    x_values = [1, 100, 500, 2500]
    y_values = [p[0], p[99], p[499], p[2499]]
    markers = ['s', '^', 'h', 'p']
    sizes = [60, 80, 80, 80]
    for x, y, marker, size in zip(x_values, y_values, markers, sizes):
        plt.scatter(x, y, marker=marker, s=size, c='k')

plt.legend(frameon=False, title="Zipf Exponent", fontsize=12)
plt.plot()
plt.yscale('log')
plt.xscale('log')
sns.despine()
plt.xlabel('Rank', fontsize=12)
plt.ylabel('Population', fontsize=12)
plt.tight_layout()
plt.show()
```
